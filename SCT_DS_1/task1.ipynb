```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("tested (1).csv")
print("dataset loaded successfully")
```

```python
print(df.head())
```

```python
df.info()
```

```python
print(df.head())
```

```python
#  fill missing age value
df['Age'].fillna(df['Age'].median())

# fill missing embarked value
df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
#visualisation for gender distribution
gender_counts=df["Sex"].value_counts()
plt.figure(figsize=(6,4))
sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
)
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Distribution of Gender")
plt.show()
```

```python
#visualisation of age distribution
plt.figure(figsize=(8,6))
sns.histplot(df["Age"].dropna(), bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#visualisation for passenger class distribution
class_counts=df["Pclass"].value_counts()
plt.figure(figsize=(5,6))
sns.barplot(x=class_counts.index, y=class_counts.values)
plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
#visualisation for survival distribution
survival_counts=df["Survived"].value_counts()
plt.figure(figsize=(4,6))
sns.barplot(x=survival_counts.index, y=survival_counts.values)
plt.title("Survival Distribution")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()
```

```python
#visualisation for fare distribution
plt.figure(figsize=(5,5))
sns.histplot(df["Fare"].dropna(), bins=20, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

Insights:

1. Male passengers were higher in number than female passengers.
2. Most passengers were between 20–40 years old.
3. Passenger Class 3 had the highest number of passengers.
4. Survival rate was lower overall.
5. Most passengers paid lower fares.
